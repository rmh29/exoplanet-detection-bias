```python
import sqlite3
import pandas as pd
conn = sqlite3.connect("../data/exoplanets.db")

def run(query):
    return pd.read_sql(query, conn)
```

```python
run("""
    SELECT disc_year, discoverymethod, COUNT(*) AS n
    FROM planets
    GROUP BY disc_year, discoverymethod
    ORDER BY disc_year
""")
```

```python
run("""
    SELECT discoverymethod,
        COUNT(*) AS n,
        ROUND(100.0 * SUM(pl_rade IS NULL) / COUNT(*), 1) AS pct_missing_radius,
        ROUND(100.0 * SUM(pl_bmasse_measured IS NULL) / COUNT(*), 1) AS pct_missing_measured_mass
    FROM planets
    GROUP BY discoverymethod
    ORDER BY n DESC
""")
```

```python
run("""
    SELECT discoverymethod, pl_bmassprov, COUNT(*) AS n
    FROM planets
    GROUP BY discoverymethod, pl_bmassprov
    ORDER BY n DESC
""")
```
